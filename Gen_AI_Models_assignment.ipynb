```python
# Build a Generative AI Assistant
## Name: Prisha Kanikaram

## Business scenario:** I am a marketing assistant for HydraSmart, a company that sells a
## Smart Water Bottle. I use three pretrained AI models to summarize a report,
##write product descriptions, and create product images.

## Part A — Text Summarization (FLAN-T5)

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Load the FLAN-T5 model
tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-base")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-base")

# My business report
report = """HydraSmart Inc. Q3 Report: Sales of the Smart Water Bottle rose 18% compared
to Q2, driven mainly by college students and gym members. Online sales made up 65%
of total revenue. Customer complaints increased about battery life, which averages
4 days instead of the advertised 7 days. The company plans to release a longer-lasting
battery in Q1 and expand into retail stores next year."""

# Ask the model to summarize
inputs = tokenizer("summarize: " + report, return_tensors="pt")
output = model.generate(**inputs, max_new_tokens=60)
summary = tokenizer.decode(output[0], skip_special_tokens=True)

print("ORIGINAL REPORT:\n")
print(report)
print("\nGENERATED SUMMARY:\n")
print(summary)
```

```python
## Part B — Text Generation (GPT-2)

from transformers import pipeline, set_seed

# Load GPT-2
generator = pipeline("text-generation", model="openai-community/gpt2")
set_seed(42)  # makes results repeatable
print("GPT-2 loaded successfully!")

# Prompt 1: short and vague
prompt1 = "The Smart Water Bottle helps busy people"

result1 = generator(
    prompt1,
    max_new_tokens=60,
    do_sample=True,
    temperature=0.8,
    top_k=50,
    pad_token_id=generator.tokenizer.eos_token_id
)
print("PROMPT 1 OUTPUT:\n")
print(result1[0]["generated_text"])
```

```python
# Prompt 2: more detailed, written like the start of an ad
prompt2 = "Introducing the Smart Water Bottle: a sleek, leak-proof bottle that tracks how much water you drink and"

result2 = generator(
    prompt2,
    max_new_tokens=60,
    do_sample=True,
    temperature=0.8,
    top_k=50,
    pad_token_id=generator.tokenizer.eos_token_id
)
print("PROMPT 2 OUTPUT:\n")
print(result2[0]["generated_text"])
```

```python
## Part C — Image Generation (Stable Diffusion)

import torch
from diffusers import StableDiffusionPipeline
import matplotlib.pyplot as plt

# Load Stable Diffusion onto the GPU
pipe = StableDiffusionPipeline.from_pretrained(
    "stable-diffusion-v1-5/stable-diffusion-v1-5",
    torch_dtype=torch.float16
).to("cuda")

img_prompt1 = "a water bottle"
img_prompt2 = ("a sleek matte black smart water bottle with a glowing LED display, "
               "professional product photo, white background, studio lighting")

seed = torch.Generator("cuda").manual_seed(42)  # repeatable results
image1 = pipe(img_prompt1, generator=seed).images[0]
image2 = pipe(img_prompt2, generator=seed).images[0]

# Show both images side by side
fig, axes = plt.subplots(1, 2, figsize=(12, 6))
axes[0].imshow(image1); axes[0].set_title("Prompt 1: simple"); axes[0].axis("off")
axes[1].imshow(image2); axes[1].set_title("Prompt 2: detailed"); axes[1].axis("off")
plt.show()

```
